## Compare within category

Use total return after fees, rolling volatility and drawdown, portfolio holdings, duration/credit quality where relevant, concentration, benchmark appropriateness, tracking difference for index funds, AUM, subscription/redemption terms, and tax treatment. Do not compare an equity fund with a money-market fund using return alone.
