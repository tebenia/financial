## Research checklist

1. Download the official memorandum/prospectus and record retrieval date.
2. Identify issuer, series, maturity, coupon/imbalan formula, payment dates, tradability, early-redemption/call terms, tax, minimum order, and settlement.
3. For traded bonds calculate dirty price, accrued interest, yield to maturity, duration, and plausible sale spread.
4. For corporate debt analyze rating, seniority, covenants, refinancing, and default recovery—not just the coupon.
