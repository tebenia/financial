## Gate before any allocation

Verify the current Indonesian regulator and authorized-provider status, asset eligibility, custody arrangement, withdrawal test, fees/spread, wallet/address controls, tax, and incident history. Understand the asset independently of the platform. Never paste seed phrases, private keys, API secrets, or account identifiers into this project.

```python
risk_questions = [
    'What contractual cash flow or economic service supports value?',
    'Who holds the private keys?',
    'Can the asset be withdrawn on-chain?',
    'What happens if the platform becomes insolvent?',
    'What is the maximum loss in IDR and can I survive it?',
    'Which current official rule and authorized-provider list did I verify?'
]
risk_questions
```
