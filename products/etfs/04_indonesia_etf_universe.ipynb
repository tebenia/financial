## Minimum pre-trade study

Record exchange status, latest prospectus, index methodology, AUM, management fee, tracking difference, indicative NAV if available, bid, ask, spread percentage, depth at several levels, normal daily traded value, creation-unit mechanism, and tax/fees. Observe several trading days; a one-minute snapshot is weak evidence.
